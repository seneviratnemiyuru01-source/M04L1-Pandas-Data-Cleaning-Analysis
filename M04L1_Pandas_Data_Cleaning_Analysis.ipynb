```python
import pandas as pd

df = pd.read_csv('m4_ecommerce_orders.csv')

print(df.shape)
print(df.dtypes)
print(df.isna().sum())
print(df.describe())
```

```python
# Missing values
print(df.isna().sum())

# Impossible values
print("Negative shipping costs:", (df['shipping_cost'] < 0).sum())
print("Quantities below one:", (df['quantity'] < 1).sum())
print("Negative unit prices:", (df['unit_price'] < 0).sum())
```

```python
before = df['quantity'].mean()
after = df.dropna(subset=['unit_price'])['quantity'].mean()

print("Before:", before)
print("After:", after)
```

```python
df_clean = df.dropna(subset=['unit_price']).copy()

df_clean = df_clean[df_clean['shipping_cost'] >= 0].copy()

print(df_clean.shape)
```

| Problem | WHAT | DID | COSTS |
|---|---|---|---|
| Missing unit price | 21 missing prices | Removed 21 records | Lost 21 orders and potentially useful information |
| Negative shipping cost | 13 negative shipping values | Removed 13 records | Lost 13 orders that might have been useful |

### Step 4: Average Quantity by Category

```python

print(
    df_clean.groupby('category')['quantity']
    .mean()
    .sort_values(ascending=False)
)
```

The category with the highest average quantity per order is Toys, with an average of 5.3125.

### Average Shipping Cost by Region


```python

print(
    df_clean.groupby('region')['shipping_cost']
    .mean()
    .sort_values(ascending=False)
)
```

The region with the highest average shipping cost is South, with an average of 13.307869.
